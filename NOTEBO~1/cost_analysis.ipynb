# Cost Model Analysis

Estimate monthly Azure spend based on traffic assumptions.

```python
# Assumptions
monthly_queries = 50_000
avg_input_tokens = 800
avg_output_tokens = 400

# GPT-4o pricing (approximate)
price_input  = 0.0025 / 1000  # per token
price_output = 0.010  / 1000  # per token

llm_cost = monthly_queries * (avg_input_tokens * price_input + avg_output_tokens * price_output)
print(f"LLM cost/month: ${llm_cost:,.0f}")
```
