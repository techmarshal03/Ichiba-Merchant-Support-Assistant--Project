# RAG Pipeline Evaluation

Evaluate hybrid RAG retrieval quality on the golden dataset.

```python
import json
import asyncio
from pathlib import Path

# Load golden dataset
with open("../tests/eval/golden_dataset_ja.jsonl") as f:
    jp_samples = [json.loads(l) for l in f]

print(f"JP samples: {len(jp_samples)}")
```

```python
# Run retrieval and measure recall@5
async def evaluate_retrieval():
    from backend.rag.retriever import hybrid_rag_retrieve
    hits = 0
    for sample in jp_samples:
        chunks = await hybrid_rag_retrieve(sample["query"], sample["domain"], "ja")
        if any(sample["expected_keywords"][0] in c.content for c in chunks):
            hits += 1
    print(f"Recall@5: {hits}/{len(jp_samples)} = {hits/len(jp_samples):.2%}")

asyncio.run(evaluate_retrieval())
```
